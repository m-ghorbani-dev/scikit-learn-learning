# Feature Extraction

## Learning objectives
- Extract calendar features from dates.
- Convert text to numeric features using count and TF-IDF vectorizers.
- Distinguish feature extraction from feature selection.

```python
import pandas as pd
df = pd.DataFrame({
    "OrderDate": pd.to_datetime(["2025-01-03", "2025-02-14", "2025-03-21", "2025-07-04", "2025-11-28", "2025-12-31"]),
    "Review": ["Great product and fast delivery", "The quality is excellent", "Delivery was late", "Good value for money", "Poor packaging but useful", "Excellent product"]
})
df
```

## 1. Extract date components

```python
date_features = pd.DataFrame(index=df.index)
date_features["year"] = df["OrderDate"].dt.year
date_features["month"] = df["OrderDate"].dt.month
date_features["day_of_week"] = df["OrderDate"].dt.dayofweek
date_features["day_of_year"] = df["OrderDate"].dt.dayofyear
date_features["is_weekend"] = (df["OrderDate"].dt.dayofweek >= 5).astype(int)
date_features
```

## 2. CountVectorizer
Each row is represented by counts of vocabulary terms.

```python
from sklearn.feature_extraction.text import CountVectorizer
count_vectorizer = CountVectorizer()
counts = count_vectorizer.fit_transform(df["Review"])
count_df = pd.DataFrame(counts.toarray(), columns=count_vectorizer.get_feature_names_out())
count_df
```

## 3. TF-IDF
TF-IDF downweights terms that appear in many documents.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
tfidf = tfidf_vectorizer.fit_transform(df["Review"])
pd.DataFrame(tfidf.toarray(), columns=tfidf_vectorizer.get_feature_names_out()).round(3)
```

## Key takeaways
- Feature extraction creates a representation from raw data.
- Feature selection keeps a subset of existing features.
- Fit text vocabularies on training text only in predictive workflows.
