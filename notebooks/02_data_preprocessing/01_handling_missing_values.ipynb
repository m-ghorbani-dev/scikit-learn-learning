# Handling Missing Values

## Learning objectives
- Inspect missing data.
- Impute numeric and categorical features.
- Understand why preprocessing should be fitted on training data only.

This notebook uses a small built-in example and needs no downloads.

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 35, np.nan, 45, 28, np.nan, 50],
    "Fare": [7.25, 12.00, 8.05, np.nan, 21.50, 9.00, 30.00],
    "Embarked": ["S", "C", "S", "Q", np.nan, "S", "C"],
    "Survived": [0, 1, 0, 1, 1, 0, 1],
})
df
```

## 1. Inspect missing values
`isna()` identifies missing entries; summing its Boolean results counts them per column.

```python
df.isna().sum()
```

```python
df.isna().mean().mul(100).round(2)
```

## 2. Separate features and target

```python
X = df.drop(columns="Survived")
y = df["Survived"]
print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

## 3. Impute numeric features
`SimpleImputer(strategy="median")` replaces missing values with each column's median. In a predictive workflow, fit it on training data only.

```python
from sklearn.impute import SimpleImputer

numeric_features = ["Age", "Fare"]
numeric_imputer = SimpleImputer(strategy="median")
X_numeric = pd.DataFrame(
    numeric_imputer.fit_transform(X[numeric_features]),
    columns=numeric_features, index=X.index
)
X_numeric
```

## 4. Impute categorical features
The most frequent category is one option; a constant label such as `Unknown` can be useful when missingness itself matters.

```python
categorical_features = ["Embarked"]
mode_imputer = SimpleImputer(strategy="most_frequent")
X_categorical = pd.DataFrame(
    mode_imputer.fit_transform(X[categorical_features]),
    columns=categorical_features, index=X.index
)
X_categorical
```

```python
constant_imputer = SimpleImputer(strategy="constant", fill_value="Unknown")
X_unknown = pd.DataFrame(
    constant_imputer.fit_transform(X[categorical_features]),
    columns=categorical_features, index=X.index
)
X_unknown
```

## Key takeaways
- Inspect missingness before choosing a strategy.
- Median imputation is often useful for numeric features.
- Use most-frequent or constant imputation for categorical features when appropriate.
- Fit imputers on training data only to avoid data leakage.
