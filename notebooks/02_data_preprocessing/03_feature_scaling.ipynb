# Feature Scaling

## Learning objectives
- Apply standardization and min-max scaling.
- Fit scalers on training data only.
- Understand which models often benefit from scaling.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [18, 25, 32, 40, 55, 63],
    "AnnualIncome": [18000, 32000, 48000, 62000, 95000, 120000],
    "SpendingScore": [72, 65, 80, 55, 40, 35],
})
df
```

## 1. Standardization
`StandardScaler` subtracts the training mean and divides by the training standard deviation.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_train, X_test = train_test_split(df, test_size=0.33, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
scaled_df
```

```python
print("Training means:")
print(scaled_df.mean().round(6))
print("\nTraining standard deviations:")
print(scaled_df.std(ddof=0).round(6))
```

## 2. Min-max scaling
`MinMaxScaler` usually maps each feature to the range 0–1. Test values may fall outside that range if they exceed the training range.

```python
from sklearn.preprocessing import MinMaxScaler

minmax = MinMaxScaler()
train_minmax = minmax.fit_transform(X_train)
test_minmax = minmax.transform(X_test)
pd.DataFrame(train_minmax, columns=X_train.columns, index=X_train.index)
```

## Why split before fitting?
A scaler learns statistics from data. Fit it on training data, then use only `transform` on test data to prevent information leakage.

```python
pd.Series(scaler.mean_, index=df.columns, name="Training mean")
```

## Key takeaways
- Standardization is often useful for linear models, SVMs, and distance-based methods.
- Min-max scaling maps features to a chosen range.
- Tree-based models usually do not require scaling.
- Use `fit_transform` on training data and `transform` on test data.
