# 2. Features, Target, and DataFrames

## Learning objectives
- Separate features (`X`) and target (`y`).
- Understand dataset shape.
- Inspect data types and missing values.


## 1. Load the example dataset

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)
df = iris.frame
display(df.head())
print("Rows and columns:", df.shape)
```

## 2. Identify features and target

```python
target_column = "target"
X = df.drop(columns=target_column)
y = df[target_column]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Feature columns:", X.columns.tolist())
```

## 3. Inspect data types and missing values

```python
print("Data types:")
display(X.dtypes.to_frame("dtype"))
print("Missing values per column:")
display(df.isna().sum().to_frame("missing_values"))
```

```python
print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])
display(y.head())
display(y.value_counts().sort_index().rename(index=dict(enumerate(iris.target_names))))
```
