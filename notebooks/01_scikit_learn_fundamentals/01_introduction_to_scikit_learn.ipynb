# 1. Introduction to Scikit-learn

## Learning objectives
- Understand what Scikit-learn is used for.
- Identify common machine learning workflow stages.
- Load and inspect a built-in dataset.

Scikit-learn provides tools for data preprocessing, model training, prediction, model selection, and evaluation. This notebook uses the built-in Iris dataset, so no download is required.


## 1. Import libraries

```python
import numpy as np
import pandas as pd
import sklearn
from sklearn.datasets import load_iris

print("Scikit-learn version:", sklearn.__version__)
```

## 2. Load the dataset

```python
iris = load_iris(as_frame=True)
X = iris.data
y = iris.target

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
```

## 3. Explore the data

```python
display(X.head())
print("Feature names:", list(X.columns))
print("Target names:", list(iris.target_names))
print("Target counts:")
display(y.value_counts().sort_index())
```

## Key concepts
- **Samples:** individual rows.
- **Features (`X`):** input variables used by a model.
- **Target (`y`):** the value to predict in supervised learning.
- **Estimator:** an object that implements a learning or transformation task.

A typical workflow is load data → prepare data → split data → train → predict → evaluate.
